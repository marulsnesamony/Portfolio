```python
# Retrieve the contents of a GitHub repository folder using the API.
# Fine grained Personal Access Token(PAT) was created in Github for this specific repo
def get_github_folder_contents(username, repository, folder_path, token=None):

    api_url = f'https://api.github.com/repos/{username}/{repository}/contents/{folder_path}'
    headers = {'Authorization': f'token {token}'} if token else {}
    
    response = requests.get(api_url, headers=headers)
    if response.status_code == 200:
        return response.json()
    else:
        print(f"Failed to fetch contents. Status code: {response.status_code}")
        return None
```

```python
# Read multiple CSV files from a GitHub folder using the API.

def read_multiple_files(username, repository, folder_path, token=None):

    contents = get_github_folder_contents(username, repository, folder_path, token)
    if not contents:
        return None
    
    
    list_dfs = []  # Storing the dataframes in a list
    i =0
    for item in contents:
        # Check if the item is a file and ends with .csv
#         i+=1
#         if i > 10:
#             return None
        if item['type'] == 'file' and item['name'].endswith('.csv') and not ("All_Years" in item['name']):
            download_url = item['download_url']
            season = item['name'][7:14]
            try:
                df = pd.read_csv(download_url)  #storing the file contents in a dataframe
                df['Season'] = season
                list_dfs.append(df)
            except Exception as e:
                print(f"Error reading {item['name']}: {e}")

    if list_dfs:
        combined_df = pd.concat(list_dfs, ignore_index=True)
        return combined_df
    else:
        return None
```
